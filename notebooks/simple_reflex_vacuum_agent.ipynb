### Problem Statement
Develop a Python program to simulate a **Simple Reflex Agent** operating in a two-square vacuum environment (Location A and Location B).
The agent must perceive the status of its current location (`Clean` or `Dirty`) and execute an action (`Suck`, `Right`, or `Left`) based strictly on current percepts without maintaining a history of past states.

### Agent PEAS Specification
| Element | Description |
| :--- | :--- |
| **P**erformance Measure | +10 points for each clean square, -1 point per action step |
| **E**nvironment | Two discrete locations: `A` and `B`, each either `Clean` or `Dirty` |
| **A**ctuators | `Suck`, `Right`, `Left`, `NoOp` |
| **S**ensors | Current location sensor (`A` or `B`), Dirt detector (`Clean` or `Dirty`) |

### Condition-Action Reflex Rules
$$\text{Action} = f(\text{Location}, \text{Status})$$

- **Rule 1**: If $\text{Status} = \text{Dirty} \implies \text{Action} = \text{Suck}$
- **Rule 2**: If $\text{Status} = \text{Clean} \land \text{Location} = A \implies \text{Action} = \text{Right}$
- **Rule 3**: If $\text{Status} = \text{Clean} \land \text{Location} = B \implies \text{Action} = \text{Left}$

```python
import sys
from pathlib import Path

# Ensure project root is on sys.path when running from notebooks/
project_root = Path("..").resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.vacuum_agent import (
    SimpleReflexVacuumAgent,
    TwoSquareEnvironment,
    GridVacuumEnvironment,
    LOCATION_A,
    LOCATION_B,
    CLEAN,
    DIRTY,
)
print("Loaded SimpleReflexVacuumAgent and Environments successfully.")
```

## 1. Simulation: Sample Scenario

- **Initial Location**: `A`
- **Location A Status**: `Dirty`
- **Location B Status**: `Clean`
- **Time Steps**: 5

```python
# Define environment states using a Python dictionary
initial_state = {
    LOCATION_A: DIRTY,
    LOCATION_B: CLEAN
}

env = TwoSquareEnvironment(initial_location=LOCATION_A, state=initial_state)
agent = SimpleReflexVacuumAgent()

print(f"Initial Environment State: {env.state}")
print(f"Agent Starting Location:  {env.agent_location}\n")
print("--- Step-by-Step Simulation Trace ---")

trace = env.simulate(steps=5, agent=agent)
for record in trace:
    print(record.format_trace())

print("\n--- Final Summary ---")
print(f"Final Environment State: {env.state}")
print(f"All Squares Clean:       {env.is_all_clean()}")
```

## 2. Test Case: Both Locations Dirty (Starting at Location B)

Validates the agent handling starting at location B with both squares dirty.

```python
env_both_dirty = TwoSquareEnvironment(
    initial_location=LOCATION_B,
    state={LOCATION_A: DIRTY, LOCATION_B: DIRTY}
)

trace_both_dirty = env_both_dirty.simulate(steps=5)
for record in trace_both_dirty:
    print(record.format_trace())

print(f"\nFinal State: {env_both_dirty.state}")
```

## 3. Demonstrating Limitations: Infinite Oscillation

Because a simple reflex agent has **no memory** of past visits or belief states, when both squares are clean, it oscillates indefinitely between A and B.

```python
env_all_clean = TwoSquareEnvironment(
    initial_location=LOCATION_A,
    state={LOCATION_A: CLEAN, LOCATION_B: CLEAN}
)

trace_oscillation = env_all_clean.simulate(steps=6)
for record in trace_oscillation:
    print(record.format_trace())
```

## 5. University Viva Voce Reference Notes

### Q1: What is a Rational Agent?
> **Answer**: A rational agent is one that for each possible percept sequence, selects an action that is expected to maximize its performance measure, given the evidence provided by the percept sequence and whatever built-in knowledge the agent has. Rationality is different from omniscience: an omniscient agent knows the *actual* outcome of its actions and can act accordingly, whereas a rational agent acts to maximize *expected* performance based on percepts received so far.

### Q2: What is the difference between Fully Observable and Partially Observable environments?
> **Answer**: In a **fully observable** environment, the agent's sensors give it access to the complete state of the environment at each point in time. In a **partially observable** environment, some aspects of the state are missing from sensor data (due to noisy or limited sensors). The 2-square vacuum cleaner environment is partially observable because the agent only senses whether its *current* square is clean or dirty; it does not know the status of the other square without physically moving to it.

### Q3: What are the main limitations of Simple Reflex Agents?
> **Answer**:
> 1. **No Memory / History**: They select actions based only on the current percept $P \to A$, completely ignoring past percepts.
> 2. **Susceptibility to Infinite Loops**: In partially observable environments, simple reflex agents often get trapped into infinite repetitive loops (e.g. bouncing between clean squares).
> 3. **Cannot Plan**: They lack internal models or goal representations, meaning they cannot anticipate future consequences or plan multi-step actions.
