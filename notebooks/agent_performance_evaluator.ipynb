### Problem Statement
Design a Python program to evaluate an autonomous delivery drone's performance.
The agent receives:
- **+10 points** for a successful delivery
- **-1 point** for every time step taken (`move`)
- **-5 points** for bumping into an obstacle (`bump`)

Simulate a log of actions and calculate the final PEAS performance score.

### Task Requirements
1. Read a sequence of simulated actions/events.
2. Maintain a running score based on the performance metric.
3. Process the log sequentially.
4. Print the final calculated score.
5. **Input Requirements**: List of string events: `['move', 'bump', 'move', 'deliver']`
6. **Expected Output**: Integer representing total performance score.
7. **Constraints / Conditions**: Score can be negative.
8. **Sample Input**: `['move', 'move', 'bump', 'move', 'deliver']`
9. **Sample Output**: `Performance Score: 2 (10 - 3 moves - 5 bump)`
10. **Optional Challenge**: Read the action log from an external CSV file.

### Theoretical Foundation: PEAS Framework for Autonomous Delivery Drone

| PEAS Dimension | Drone Specification | Impact on Agent Decision-Making |
| :--- | :--- | :--- |
| **P - Performance Measure** | $+10$ / delivery, $-1$ / move, $-5$ / bump | Objective utility function balancing speed, safety, and delivery throughput |
| **E - Environment** | Urban 3D airspace, buildings, power lines, wind, landing pads | Dynamic, partially observable, stochastic, continuous 3D world |
| **A - Actuators** | Rotors/ESCs, pitch/roll/yaw attitude motors, package winch/gripper | Translates movement actions into physical flight trajectory |
| **S - Sensors** | GPS, 3D LiDAR, downward camera, ultrasonic proximity, IMU | Provides real-time percepts for obstacle detection and landing |

### Step 5: Custom Performance Metric Sensitivity Analysis

In real AI system engineering, tuning the weights of the performance measure alters agent rationality and behavior trade-offs.

```python
# Aggressive safety-critical configuration (harsh bump penalty):
safety_first_metric = ScoringMetric(reward_delivery=10, cost_move=1, penalty_bump=20)
safe_evaluator = DronePerformanceEvaluator(metric=safety_first_metric)

# High-value parcel delivery configuration:
high_value_metric = ScoringMetric(reward_delivery=50, cost_move=2, penalty_bump=5)
high_value_evaluator = DronePerformanceEvaluator(metric=high_value_metric)

print("Standard Score:    ", evaluator.evaluate_log(sample_events).final_score)
print("Safety-First Score:", safe_evaluator.evaluate_log(sample_events).final_score)
print("High-Value Score:  ", high_value_evaluator.evaluate_log(sample_events).final_score)
```
