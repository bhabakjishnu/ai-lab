### Problem Statement
Write a Python program that classifies AI problems based on their environment characteristics.
The user inputs binary answers for **Observability**, **Determinism**, **Episodic nature**, **Statics**, and **Discreteness**.
The program maps these to known AI environments (e.g., Chess, Poker, Medical Diagnosis).

### Task Requirements
1. Define a knowledge dictionary mapping boolean tuples to environment names.
2. Prompt the user for 5 environment properties.
3. Match the input tuple to the dictionary in $O(1)$ time.
4. Handle unknown combinations gracefully.
5. **Sample Input**: `True, False, False, False, True` (Observable, Stochastic, Sequential, Dynamic, Discrete)
6. **Sample Output**: `Environment Match: Backgammon`
7. **Optional Challenge**: Use a distance metric (Hamming distance) to find the closest environment if an exact match fails.

### Theoretical Foundation: The 5 Environmental Dimensions (Russell & Norvig AIMA)

| Dimension | True State (Affirmative) | False State (Negative) | Key Distinction |
| :--- | :--- | :--- | :--- |
| **1. Observability** | **Fully Observable** | **Partially Observable** | Sensors provide complete world state vs. sensor noise/hidden state |
| **2. Determinism** | **Deterministic** | **Stochastic** | Next state uniquely determined by action vs. governed by chance/dice |
| **3. Episodic Nature** | **Episodic** | **Sequential** | Current action independent of past/future vs. current actions dictate future states |
| **4. Statics** | **Static** | **Dynamic** | Environment static during deliberation vs. changes continuously with time |
| **5. Discreteness** | **Discrete** | **Continuous** | Finite distinct states/actions/time vs. real-valued smooth physical variables |

### Step 3: Classifying Benchmark AI Environments
Evaluating classic textbook benchmarks: Chess (with/without clock), Poker, Medical Diagnosis, Taxi Driving, and Part-Picking Robot.

```python
test_cases = [
    ("Chess (with clock)", (True, True, False, False, True)),
    ("Chess (without clock)", (True, True, False, True, True)),
    ("Poker (timed)", (False, False, False, False, True)),
    ("Poker (untimed)", (False, False, False, True, True)),
    ("Medical Diagnosis / Taxi Driving", (False, False, False, False, False)),
    ("Part-Picking Robot", (False, False, True, False, False)),
    ("Image Analysis (Discrete)", (True, True, True, True, True)),
]

for label, props in test_cases:
    res = classifier.classify(props)
    print(f"[{label}]:")
    print(f"   Input: {props}")
    print(f"   Match: {res.matched_environment}")
    print(f"   PEAS Environment: {res.metadata.peas.environment}")
    print(f"   PEAS Performance: {res.metadata.peas.performance_measure}\n")
```

### Step 4: Graceful Handling of Unknown Combinations
When a user inputs a combination that does not have an exact textbook match in the knowledge dictionary, the system handles it gracefully without throwing an unhandled exception.

```python
unknown_input = (True, False, True, False, True)
res_exact_only = classifier.classify(unknown_input, allow_nearest=False)

print(f"Exact match found? {res_exact_only.is_exact_match}")
print(f"Matched Environment: {res_exact_only.matched_environment}")
print(f"Graceful Message:   {res_exact_only.message}")
```

### Step 5: Optional Challenge — Nearest-Neighbor Hamming Distance Metric
When an exact match is missing, we compute the Hamming distance $D_H(u, v) = \sum_{i=0}^4 (u_i \neq v_i)$ across all entries in the knowledge base.
This identifies the closest environment and explains exactly which properties differ.

```python
res_challenge = classifier.classify(unknown_input, allow_nearest=True)
print(format_classification_report(res_challenge))
```

### Step 7: Viva Voce Reference & Study Guide

1. **PEAS Framework**:
   - **Performance Measure**: External objective metric for success.
   - **Environment**: External world context.
   - **Actuators**: Mechanisms for agent actions.
   - **Sensors**: Devices collecting percepts.

2. **Chess vs. Poker**:
   - Chess: Fully Observable, Deterministic, Sequential, Static/Semi-dynamic, Discrete, Multi-Agent (Perfect Info).
   - Poker: Partially Observable, Stochastic, Sequential, Dynamic/Static, Discrete, Multi-Agent (Imperfect Info, Bluffing).

3. **Continuous vs. Discrete**:
   - Discrete: Finite distinct states and actions (Chess squares, grid cells).
   - Continuous: Infinite real-valued ranges (Steering angle, velocity, medical vitals).

4. **Lookup Table Complexity**:
   - Time Complexity: $O(1)$ dictionary lookup.
   - Space Complexity: $O(K)$ entries, where $K \le 32$ total boolean combinations.
