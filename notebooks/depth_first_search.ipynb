### 1. Theoretical Foundations & Algorithmic Principles

#### A. Last-In, First-Out (LIFO) Frontier
- DFS explores down a single branch to its deepest leaf before backtracking.
- In **Iterative DFS**, an explicit Python `list` acts as a LIFO stack:
  - `stack.append(node)` pushes in $O(1)$ amortized time.
  - `stack.pop()` removes the top element in $O(1)$ time.
- In **Recursive DFS**, the operating system / Python runtime call stack serves as the implicit LIFO stack.

#### B. The Traversal Ordering Nuance (Iterative vs. Recursive)
- **Iterative Forward-Push (Sample Output Match)**:
  When node `'1'` pushes neighbors `['2', '3']` in order onto the stack, `'3'` ends up at the top.
  Popping `'3'` first yields traversal: **`1, 3, 2, 4`** (matches the university question sample output!).
- **Standard Recursive (Left-to-Right)**:
  Exploring `'2'` first down to `'4'`, then backtracking to `'3'`, yields traversal: **`1, 2, 4, 3`**.
- **Order-Preserving Iterative (Reverse Push)**:
  Pushing neighbors in reversed order `['3', '2']` places `'2'` on top, producing identical order to standard recursion (**`1, 2, 4, 3`**).

#### C. Space Complexity Advantage over BFS
- **BFS**: Requires $O(b^d)$ exponential memory to hold entire frontier levels.
- **DFS**: Only needs to store the active branch path plus unexpanded siblings at each depth level ($O(b \cdot m)$ linear space for tree search, $O(V)$ for graph search).

### 7. Viva Voce Reference & Examination Guide

Key viva topics required by the laboratory syllabus:
- **Stack vs Queue**: DFS uses LIFO stack (last-added is explored first, driving deep into branches); BFS uses FIFO queue (level-by-level).
- **Why DFS is not optimal**: DFS commits to the first branch it sees. A deep path of length 1,000 will be taken before exploring a shallow 1-step goal on an alternate branch.
- **Space complexity advantage**: DFS stores only the current branch path $O(b \cdot m)$, whereas BFS stores the entire level $O(b^d)$ (exponential memory explosion).
