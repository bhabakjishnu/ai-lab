### 1. Theoretical Foundation & Algorithm Mechanics

#### Why `collections.deque` Over Standard Python `list`?
- A Python `list` is implemented as dynamic array of pointers. Dequeueing from the front with `list.pop(0)` takes $O(N)$ time because all remaining elements must shift left in memory.
- `collections.deque` is implemented in CPython as a doubly-linked list of fixed-size memory blocks. Enqueueing via `append()` and dequeueing via `popleft()` both execute in $O(1)$ amortized time.

#### Time & Space Complexity Derivation
- **Time Complexity**: $O(V + E)$
  - Every reachable vertex $v \in V$ is enqueued and dequeued at most once: $O(V)$.
  - During expansion of vertex $u$, all its outgoing edges $(u, v) \in E$ are evaluated: $\sum_{u \in V} \text{deg}^+(u) = O(E)$.
- **Space Complexity**: $O(V)$
  - The `visited` set stores at most $V$ vertex labels.
  - The `frontier` deque holds at most the maximum breadth of any level, bounded by $V$.

### 5. BFS Spanning Tree & Shortest Path Reconstruction

In an unweighted graph, BFS produces the shortest path (minimum edge hops) between the source and any reachable vertex.

````python
print("Hierarchical BFS Spanning Tree:")
print(result.format_ascii_tree())
print("\nShortest Paths from 'A':")
for target in sorted(result.graph.keys()):
    path = result.shortest_path_to(target)
    path_str = " -> ".join(path) if path else "Unreachable"
    hops = len(path) - 1 if path else -1
    print(f"  • A to {target:<2}: {path_str:<20} (Hops: {hops})")
````
