### 1. Theoretical Foundations: Why $O(b^{d/2})$ Massively Outperforms $O(b^d)$

| Search Strategy | Frontier Type | Time Complexity | Space Complexity | Optimality (Unit Cost) |
| :--- | :--- | :--- | :--- | :--- |
| **Unidirectional BFS** | Single FIFO Queue | $O(b^d)$ | $O(b^d)$ | **Guaranteed** |
| **Bidirectional BFS** | Dual FIFO Queues | **$O(b^{d/2})$** | **$O(b^{d/2})$** | **Guaranteed** |

#### The Geometry of Search Space Halving
A standard forward BFS searches a hypersphere of radius $d$ around the initial state, exploring roughly:
$$N_{\text{BFS}} = b^0 + b^1 + b^2 + \dots + b^d \approx O(b^d)$$

In contrast, Bidirectional Search grows two simultaneous hyperspheres of radius $d/2$:
$$N_{\text{BDS}} = 2 \times \left(b^0 + b^1 + \dots + b^{d/2}\right) \approx O(b^{d/2})$$

> **Concrete Numerical Comparison** ($b = 10, d = 10$):
> - **Unidirectional BFS**: $10^{10} = 10,000,000,000$ states (10 billion nodes!). Requires terabytes of RAM and hours/days of compute.
> - **Bidirectional BFS**: $2 \times 10^5 = 200,000$ states! Runs in milliseconds with a few megabytes of RAM.
> - **Speedup**: **50,000x faster execution and memory savings!**
