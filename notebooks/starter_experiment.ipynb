## 2. Data Acquisition & Experiment Simulation

We simulate a synthetic dataset representing model evaluations under varying test conditions.

```python
random.seed(42)

def generate_experiment_run(n_samples: int = 100) -> List[Dict[str, float]]:
    """Generate synthetic experiment metrics across iterations."""
    records = []
    base_accuracy = 0.65
    for epoch in range(1, n_samples + 1):
        # Simulated learning curve with noise
        progress = 1.0 - math.exp(-epoch / 25.0)
        noise = random.gauss(0, 0.02)
        acc = min(0.99, max(0.50, base_accuracy + 0.30 * progress + noise))
        loss = max(0.05, 1.2 * math.exp(-epoch / 20.0) + random.uniform(-0.02, 0.02))
        records.append({
            "epoch": epoch,
            "accuracy": round(acc, 4),
            "loss": round(loss, 4)
        })
    return records

experiment_data = generate_experiment_run(n_samples=50)
print(f"Generated {len(experiment_data)} epochs of experiment data.")
print("Sample of first 3 records:", experiment_data[:3])
```

## 3. Metric Computation & Statistical Verification

Compute key aggregate metrics including minimum loss, peak accuracy, and convergence rates.

```python
accuracies = [r["accuracy"] for r in experiment_data]
losses = [r["loss"] for r in experiment_data]

summary_stats = {
    "initial_accuracy": accuracies[0],
    "final_accuracy": accuracies[-1],
    "max_accuracy": max(accuracies),
    "min_loss": min(losses),
    "final_loss": losses[-1],
    "mean_accuracy": round(sum(accuracies) / len(accuracies), 4)
}

print("Experiment Summary Statistics:")
for metric, value in summary_stats.items():
    print(f"  - {metric}: {value}")
```

## 4. Visualizing Experiment Progression

Render an inline visualization of the training trajectory and loss decay across epochs. Both ASCII terminal charts and Matplotlib curves (if installed) are supported.

```python
try:
    import matplotlib.pyplot as plt

    epochs = [r["epoch"] for r in experiment_data]
    fig, ax1 = plt.subplots(figsize=(8, 4))

    color = "tab:blue"
    ax1.set_xlabel("Epoch")
    ax1.set_ylabel("Accuracy", color=color)
    ax1.plot(epochs, accuracies, color=color, linewidth=2, label="Accuracy")
    ax1.tick_params(axis="y", labelcolor=color)

    ax2 = ax1.twinx()
    color = "tab:red"
    ax2.set_ylabel("Loss", color=color)
    ax2.plot(epochs, losses, color=color, linestyle="--", linewidth=2, label="Loss")
    ax2.tick_params(axis="y", labelcolor=color)

    plt.title("Simulated Model Training Trajectory")
    fig.tight_layout()
    plt.show()
except ImportError:
    # Graceful fallback to formatted text/ASCII visualization when matplotlib is not installed
    print("Matplotlib not detected; rendering formatted progress table:")
    print(f"{'Epoch':<8}{'Accuracy':<12}{'Loss':<12}{'Trajectory'}")
    for r in experiment_data[::5]:
        bars = int(r['accuracy'] * 20)
        bar_str = "█" * bars + "░" * (20 - bars)
        print(f"{r['epoch']:<8}{r['accuracy']:<12.4f}{r['loss']:<12.4f}[{bar_str}]")
```

### Data Analysis Key Findings
- **Convergence Behavior**: The simulated training trajectory achieves steady asymptotic convergence, moving from an initial baseline accuracy of ~0.65 to a peak accuracy above 0.90 within 35 epochs.
- **Loss Decay**: Initial loss drops sharply from >1.0 down to a stabilized minimum loss plateau below 0.15.
