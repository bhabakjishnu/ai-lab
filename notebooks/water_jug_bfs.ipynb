### Problem Statement
Solve the classic **Water Jug Problem** using Breadth-First Search. Given two jugs of capacities $X$ and $Y$ liters, and an infinite water supply and sink, find the sequence of operations to measure exactly $Z$ liters in one of the jugs.

### Task Requirements
1. **Represent states** as tuples `(jug1_amt, jug2_amt)` where $0 \le \text{jug1} \le X$ and $0 \le \text{jug2} \le Y$.
2. **Generate successors for 6 rules**:
   - Rule 1: `Fill 1` - Fill Jug 1 to capacity $X$: $(X, y)$
   - Rule 2: `Fill 2` - Fill Jug 2 to capacity $Y$: $(x, Y)$
   - Rule 3: `Empty 1` - Empty Jug 1 completely: $(0, y)$
   - Rule 4: `Empty 2` - Empty Jug 2 completely: $(x, 0)$
   - Rule 5: `Pour 1->2` - Pour Jug 1 into Jug 2 until full or empty: $(x - d, y + d)$ where $d = \min(x, Y - y)$
   - Rule 6: `Pour 2->1` - Pour Jug 2 into Jug 1 until full or empty: $(x + d, y - d)$ where $d = \min(y, X - x)$
3. **Use BFS to find the shortest action sequence**.
4. **Return the sequence of states** from `(0,0)` to goal state.
5. **Optional Challenge**: Algebraically verify reachability before searching via Bézout's Identity.

### 1. Theoretical Foundations: State-Space Formulation & Bézout's Identity

#### Mathematical State Space
Every configuration is represented by a pair $(x, y) \in [0, X] \times [0, Y]$ of integers.
- **Initial State**: $s_0 = (0, 0)$
- **Goal Test**: $x = Z$ (or $y = Z$, or $x = Z \lor y = Z$)
- **State Space Size**: Upper bound is $(X + 1)(Y + 1)$ states. However, in practice, because operations pour until empty or full, at least one jug is always at $0$ or its maximum capacity, restricting reachable states to the perimeter and specific diagonal lines.

#### Algebraic Reachability (Bézout's Identity & GCD Rule)
Every valid operation (fill, empty, pour) modifies the total water volume by integer multiples of $X$ and $Y$:
$$\text{Volume} = a \cdot X + b \cdot Y \quad (a, b \in \mathbb{Z})$$
By **Bézout's Identity**, the greatest common divisor $\gcd(X, Y)$ is the smallest positive integer that can be written as $aX + bY$. Therefore, any linear combination $aX + bY$ is an integer multiple of $\gcd(X, Y)$.

> **Theorem (Water Jug Solvability)**:
> Target $Z$ is measurable in one of the jugs **if and only if**:
> 1. $Z \le \max(X, Y)$ (Capacity constraint)
> 2. $Z \pmod{\gcd(X, Y)} = 0$ (Divisibility constraint)
> 
> If either condition is violated, the target state is mathematically unreachable.

### 7. Optional Challenge: Algebraic Reachability & GCD Rule Analysis
Verifying reachability before executing BFS using the Extended Euclidean Algorithm and Bézout's coefficients.

```python
cases = [
    ("Canonical 4L & 3L -> 2L", 4, 3, 2, "1"),
    ("Die Hard 5L & 3L -> 4L", 5, 3, 4, "1"),
    ("Co-prime 7L & 5L -> 4L", 7, 5, 4, "1"),
    ("Unsolvable: Even Jugs 6L & 4L -> Odd Target 5L", 6, 4, 5, "any"),
    ("Unsolvable: Target 7L Exceeds Capacity 5L", 3, 5, 7, "any"),
]

for label, x, y, z, tj in cases:
    reach = check_reachability_algebraic(x, y, z, target_jug=tj)
    status = "REACHABLE" if reach.is_reachable else "UNREACHABLE"
    print(f"[{status}] {label}")
    print(f"  • gcd({x}, {y}) = {reach.gcd_value} | Bézout: {reach.bezout_x}*({x}) + {reach.bezout_y}*({y}) = {reach.gcd_value}")
    print(f"  • Reasoning: {reach.reason}\n")
```
