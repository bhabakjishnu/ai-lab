### 1. Theoretical Foundations & AI Principles

#### Why Depth-Limited Search?
Standard **Depth-First Search (DFS)** is **incomplete** in infinite state spaces or finite spaces with cycles. DFS ventures blindly down the first branch; if that branch is infinite (e.g., continuous navigation or infinite game trees), DFS gets trapped forever and never backtracks, missing shallow goals located on adjacent branches.

**Depth-Limited Search (DLS)** eliminates this failure mode by imposing a predetermined depth ceiling $l$. All nodes at depth $l$ are treated as leaf nodes with no successors, guaranteeing finite termination.

#### Tri-State Outcome Model
Unlike standard binary searches (Found vs. Not Found), DLS produces three semantic outcomes:
1. **Success**: Goal found at depth $d \le l$. Returns the path from start to goal.
2. **Cutoff**: The goal was **not** found within depth $l$, but at least one branch was truncated because the depth budget was exhausted ($limit = 0$). The goal *might* exist deeper ($d > l$).
3. **Failure**: The entire reachable search space was completely explored to natural dead ends without hitting any depth cutoffs. We have proved with certainty that the goal does **not** exist in this connected component.

#### Foundational Role in Iterative Deepening Search (IDS)
The distinction between **Cutoff** and **Failure** is the critical signal used by **Iterative Deepening Search (IDS)**:
- If DLS returns **Cutoff**, IDS increments the limit ($l \leftarrow l + 1$) and searches again.
- If DLS returns **Failure**, IDS halts immediately because searching deeper cannot discover nonexistent states.
- If DLS returns **Success**, IDS halts and returns the optimal shallowest solution.

### 6. Viva Voce Reference & Theory Guide

| Concept | Standard DFS | Depth-Limited Search (DLS) | Iterative Deepening Search (IDS) |
| :--- | :--- | :--- | :--- |
| **Time Complexity** | $O(b^m)$ | $O(b^l)$ | $O(b^d)$ |
| **Space Complexity** | $O(b \times m)$ | $O(b \times l)$ | $O(b \times d)$ |
| **Completeness** | No (fails on infinite branches/loops) | No (fails if $l < d$) | **Yes** (if $b$ is finite) |
| **Optimality** | No | No | **Yes** (for uniform step costs) |

#### Essential Viva Questions & Authoritative Answers

1. **Infinite State Spaces**:
   - *Question*: How does DLS handle infinite state spaces where standard DFS fails?
   - *Answer*: Standard DFS can plunge down an infinite path forever ($m = \infty$), never backtracking. DLS establishes a strict maximum depth limit $l$. Once depth $l$ is reached, the search halts and backtracks, guaranteeing finite execution.

2. **Cutoff vs. Failure**:
   - *Question*: What is the precise distinction between Cutoff and Failure in DLS?
   - *Answer*: **Cutoff** indicates the goal was not found within depth $l$, but at least one node at depth $l$ was truncated before expanding its children (the goal might exist deeper). **Failure** indicates the entire reachable state space was completely traversed to dead ends without hitting any depth limit bounds, definitively proving that the goal does not exist in the connected component.

3. **Importance for Iterative Deepening Search (IDS)**:
   - *Question*: Why is the tri-state outcome of DLS vital for IDS?
   - *Answer*: IDS iteratively invokes DLS with limits $l = 0, 1, 2, \dots$. If DLS returns Cutoff, IDS knows it must increase $l$. If DLS returns Failure, IDS halts immediately knowing searching deeper will never find the goal. If DLS returns Success, the optimal solution has been found.
