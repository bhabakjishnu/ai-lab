## 1. Theoretical Foundations & Distance Metrics in AI

In informed (heuristic) search algorithms like **A\*** and **Greedy Best-First Search**, the distance metric chosen as the heuristic evaluation function $h(n)$ must mirror the physical or topological constraints of the search space:

1. **Manhattan Distance ($L_1$ Norm / Taxicab Metric)**:
   $$d_M(p_1, p_2) = |x_1 - x_2| + |y_1 - y_2|$$
   Used when movement is strictly restricted to 4 orthogonal cardinal directions (Up, Down, Left, Right).

2. **Euclidean Distance ($L_2$ Norm / Straight-Line Metric)**:
   $$d_E(p_1, p_2) = \sqrt{(x_1 - x_2)^2 + (y_1 - y_2)^2}$$
   Straight-line distance 'as the crow flies'. Used for continuous space, any-angle pathfinding, or as a relaxed lower-bound heuristic.

3. **Chebyshev Distance ($L_\infty$ Norm / Chessboard Metric - Optional Challenge)**:
   $$d_C(p_1, p_2) = \max(|x_1 - x_2|, |y_1 - y_2|)$$
   Used on 8-directional grids where diagonal transitions cost the same as orthogonal transitions (e.g., King's moves on a chessboard).

4. **Octile Distance**:
   $$d_O(p_1, p_2) = D_1 \cdot \max(dx, dy) + (D_2 - D_1) \cdot \min(dx, dy)$$
   Used on 8-directional grids where orthogonal moves cost $D_1=1$ and diagonal moves cost $D_2=\sqrt{2}$.

## 5. Metric Space Axiom Verification

A valid distance metric $d(p_1, p_2)$ on a set $X$ must satisfy 4 mathematical axioms:
1. **Non-negativity**: $d(p_1, p_2) \ge 0$
2. **Identity of Indiscernibles**: $d(p_1, p_2) = 0 \iff p_1 = p_2$
3. **Symmetry**: $d(p_1, p_2) = d(p_2, p_1)$
4. **Triangle Inequality**: $d(p_1, p_3) \le d(p_1, p_2) + d(p_2, p_3)$

```python
p1, p2, p3 = (1, 1), (4, 5), (2, 8)

for metric_name, fn in [
    ("Manhattan Distance (L1)", manhattan),
    ("Euclidean Distance (L2)", euclidean),
    ("Chebyshev Distance (L_inf)", chebyshev),
    ("Octile Distance", octile),
]:
    report = verify_metric_space_axioms(p1, p2, p3, fn)
    status = "PASS" if report["all_satisfied"] else "FAIL"
    print(f"[{status}] {metric_name:<28}: Non-neg={report['non_negativity']['satisfied']}, Identity={report['identity_of_indiscernibles']['satisfied']}, Symmetry={report['symmetry']['satisfied']}, Triangle={report['triangle_inequality']['satisfied']}")
```
