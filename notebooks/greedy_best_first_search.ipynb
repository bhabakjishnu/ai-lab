## 1. Theoretical Foundations & Evaluation Function

In **Greedy Best-First Search (GBFS)**, node expansion is guided strictly by the heuristic estimate to the goal:

$$f(n) = h(n)$$

### Key Algorithmic Mechanics:
1. **Pure Heuristic Search**: Unlike Uniform Cost Search ($f(n) = g(n)$) or A* Search ($f(n) = g(n) + h(n)$), GBFS completely ignores the cumulative path cost $g(n)$ incurred from the start state.
2. **Priority Queue Min-Heap**: Python's `heapq` module implements a binary min-heap where elements are structured as `(h(n), tie_breaker, node_id, path)`.
3. **Expansion Criterion**: The node in the frontier with the smallest estimated distance to the goal ($h(n)$) is popped and expanded first.
4. **Optimality**: **NOT optimal**. It can eagerly pick a step with a small $h(n)$ that requires traversing an edge with massive cost.
5. **Completeness**: Complete in finite graphs with an explored set (graph search); incomplete in tree search due to potential loops.

## 3. Optional Challenge: Proving That Greedy Search Fails Optimality

To demonstrate that Greedy Best-First Search is **not optimal**, consider this counterexample graph:
- Start node **`S`** can branch to **`A`** ($h=2$, step cost $10$) or **`B`** ($h=8$, step cost $1$).
- Node **`A`** connects to **`G`** with step cost $100$ (Total path `S -> A -> G` cost = **$110$**).
- Node **`B`** connects to **`G`** with step cost $1$ (Total path `S -> B -> G` cost = **$2$**).

**What Greedy Does**:
Because $h(A) = 2 < h(B) = 8$, Greedy eagerly takes `S -> A -> G` with cost $110$, suffering a **$55\times$ suboptimality penalty**!

```python
comparison = compare_greedy_vs_optimal(
    SUBOPTIMAL_CHALLENGE_GRAPH,
    SUBOPTIMAL_CHALLENGE_HEURISTICS,
    start="S",
    goal="G",
)
print(comparison.format_report())
```
